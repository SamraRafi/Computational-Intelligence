PSO for Reinforcement Learning

```python
import random
import numpy as np
```

```python
class Particle:
    def __init__(inertia:float, c1:float, c2:float, len:int):
        self.iner = inertia
        self.c1 = c1
        self.c2 = c2

        self.X =np.zeros(len)
        self.X_Pbest = None
        self.X_Gbest =None
        self.Vx = np.zeros(len)


    def getParticleBest(self): return self.X_Pbest

    def setGlobalBest(self, gbest): self.X_Gbest = gbest

    def updateVelocity(self):
        loc = self.c1 * random.random * (self.X_Pbest - self.X)
        glb= self.c2 * random.random * (self.X_Gbest - self.X)
        self.Vx = self.Vx + loc + glb
        self.updatePosition()
        return
        
    def updatePosition(self):
        self.X = self.X + self.Vx
        return

```
